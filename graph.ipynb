```python
#1.adjancency_list
def create_adjacency_list (vertices,edges):
    adjancency_list=[]
    for i in range (vertices):
        adjancency_list.append ([])

    #extract edge 
    for i,j in edges:
        adjancency_list[i].append(j)
        adjancency_list[j].append(i)

    return adjancency_list

"""    #extract edge manually
    for edge in edges:
        x=edges[0]
        y=edges[1]
        adjancency_list[x].append(y)
        adjancency_list[y].append(x)
        return adjancency_list
"""
```

```python
vertices = 3
edges = [(0,1), (0,2)]
adj_list = create_adjacency_list(vertices, edges)


print("\nAdjacency List:")
for index, neighbors in enumerate(adj_list):
    print(index, "->", neighbors)
```

```python
# Generic Tree Implementation

class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []   # list of child nodes

    def add_child(self, child_node):
        self.children.append(child_node)

    def remove_child(self, child_node):
        if child_node in self.children:
            self.children.remove(child_node)

    def print_tree(self, level=0):
        indent = " " * (level * 4)
        print(f"{indent}- {self.data}")

        for child in self.children:
            child.print_tree(level + 1)


# Create nodes
root = TreeNode("A")

node_b = TreeNode("B")
node_c = TreeNode("C")
node_d = TreeNode("D")
node_e = TreeNode("E")
node_f = TreeNode("F")

# Build tree
root.add_child(node_b)
root.add_child(node_c)
root.add_child(node_d)

node_c.add_child(node_e)
node_c.add_child(node_f)

# Print tree
root.print_tree()
```

```python
#binary tree implementation 

class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def add_child(self, child_node, position):
        if position == "left":
            if self.left is None:
                self.left = child_node
            else:
                raise ValueError("Left child already exists")
        elif position == "right":
            if self.right is None:
                self.right = child_node
            else:
                raise ValueError("Right child already exists")
        else:
            raise ValueError("Position must be 'left' or 'right'")

    def remove_child(self, position):
        if position == "left":
            self.left = None
        elif position == "right":
            self.right = None
        else:
            raise ValueError("Position must be 'left' or 'right'")

    def print_tree(self, level=0):
        indent = " " * (level * 4)
        print(f"{indent}- {self.data}")

        if self.left:
            self.left.print_tree(level + 1)
        if self.right:
            self.right.print_tree(level + 1)

# Create nodes
root = BinaryTreeNode("A")
node_b = BinaryTreeNode("B")
node_c = BinaryTreeNode("C")
node_d = BinaryTreeNode("D")

# Build tree
root.add_child(node_b, "left")
root.add_child(node_c, "right")
node_c.add_child(node_d, "left")

# Print tree
root.print_tree()
```

```python
from collections import deque

class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    # ------------------------
    # Add Child
    # ------------------------
    def add_child(self, child_node, position):
        if position == "left":
            self.left = child_node
        elif position == "right":
            self.right = child_node
        else:
            raise ValueError("Position must be 'left' or 'right'")

    # ------------------------
    # Traversals
    # ------------------------

    # Preorder: Root → Left → Right
    def preorder(self):
        result = [self.data]

        if self.left:
            result += self.left.preorder()
        if self.right:
            result += self.right.preorder()

        return result

    # Inorder: Left → Root → Right
    def inorder(self):
        result = []

        if self.left:
            result += self.left.inorder()

        result.append(self.data)

        if self.right:
            result += self.right.inorder()

        return result

    # Postorder: Left → Right → Root
    def postorder(self):
        result = []

        if self.left:
            result += self.left.postorder()
        if self.right:
            result += self.right.postorder()

        result.append(self.data)
        return result

    # Level Order: BFS
    def level_order(self):
        result = []
        queue = deque([self])

        while queue:
            node = queue.popleft()
            result.append(node.data)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

        return result

    # ------------------------
    # Print Tree (visual)
    # ------------------------
    def print_tree(self, level=0):
        indent = " " * (level * 4)
        print(f"{indent}- {self.data}")

        if self.left:
            self.left.print_tree(level + 1)
        if self.right:
            self.right.print_tree(level + 1)

# Create nodes
root = BinaryTreeNode("A")
b = BinaryTreeNode("B")
c = BinaryTreeNode("C")
d = BinaryTreeNode("D")
e = BinaryTreeNode("E")
f = BinaryTreeNode("F")

# Build tree
root.add_child(b, "left")
root.add_child(c, "right")

b.add_child(d, "left")
b.add_child(e, "right")

c.add_child(f, "right")

root.print_tree()
```

## Heap implementation 

```python
#array based  heap implementation
class ArrayBinaryTree:
    def __init__(self):
        self.tree = []

    def insert(self, value):
        self.tree.append(value)

    def get_left_child(self, index):
        left = 2 * index + 1
        if left < len(self.tree):
            return self.tree[left]
        return None

    def get_right_child(self, index):
        right = 2 * index + 2
        if right < len(self.tree):
            return self.tree[right]
        return None

    def get_parent(self, index):
        if index == 0:
            return None
        parent = (index - 1) // 2
        return self.tree[parent]

    def display(self):
        print(self.tree)



tree = ArrayBinaryTree()

values = [10, 5, 7, 2, 4, 6, 8]

for val in values:
    tree.insert(val)

tree.display()

print("Left child of index 0:", tree.get_left_child(0))
print("Right child of index 0:", tree.get_right_child(0))
print("Parent of index 4:", tree.get_parent(4))
```

```python
#node  based  heap implementation
class TreeNode:
    def __init__(self, value):
        self.val = value
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    # Insert
    def insert(self, root, value):
        if root is None:
            return TreeNode(value)

        if value < root.val:
            root.left = self.insert(root.left, value)
        else:
            root.right = self.insert(root.right, value)

        return root

    # Search
    def search(self, root, value):
        if root is None or root.val == value:
            return root

        if value < root.val:
            return self.search(root.left, value)
        else:
            return self.search(root.right, value)

    # Inorder Traversal
    def inorder(self, root):
        result = []
        if root:
            result += self.inorder(root.left)
            result.append(root.val)
            result += self.inorder(root.right)
        return result

bst = BST()

values = [5, 3, 7, 2, 4, 6, 8]

for val in values:
    bst.root = bst.insert(bst.root, val)

print("Inorder Traversal:", bst.inorder(bst.root))

print("Search 4:", bst.search(bst.root, 4) is not None)
print("Search 10:", bst.search(bst.root, 10) is not None)
```

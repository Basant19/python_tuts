```python
'''
Encapsulation means wrapping data (variables) and methods (functions) together into a single unit (class) 
and restricting direct access to data.  

In Python, encapsulation is achieved using:

Public members

Protected members

Private members

| Access Level | Syntax   | Meaning                              |
| ------------ | -------- | ------------------------------------ |
| Public       | `name`   | Accessible everywhere                |
| Protected    | `_name`  | Accessible inside class & subclasses |
| Private      | `__name` | Accessible only inside the class     |

  '''
```

```python
class Student:
    def __init__(self, name):
        self.name = name   # public variable

s = Student("Basant")
print(s.name)   #  Accessible
```

```python
class Student:
    def __init__(self, name):
        self._name = name   # protected variable

class Child(Student):
    def show(self):
        print(self._name)

c = Child("Basant")
c.show()
```

```python
# Private Members
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid amount")

acc = BankAccount(1000)
print(acc.get_balance())
acc.set_balance(2000)
```

```python
# Using Property Decorators
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid amount")

acc = BankAccount(3000)
print(acc.balance)
acc.balance = 5000
```

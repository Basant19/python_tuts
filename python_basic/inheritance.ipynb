```python
# Base class
class Animal:
    def speak(self):
        print("Animal makes a sound")

# ----------------------------
# 1. Single Inheritance
# ----------------------------
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# ----------------------------
# 2. Multilevel Inheritance
# ----------------------------
class Puppy(Dog):
    def weep(self):
        print("Puppy weeps")

# ----------------------------
# 3. Hierarchical Inheritance
# ----------------------------
class Cat(Animal):
    def meow(self):
        print("Cat meows")

# ----------------------------
# 4. Multiple Inheritance
# ----------------------------
class Pet:
    def friendly(self):
        print("Pet is friendly")

class HybridAnimal(Dog, Pet):
    def show(self):
        print("HybridAnimal created")

# ----------------------------
# Testing all inheritance types
# ----------------------------

print("Single Inheritance:")
d = Dog()
d.speak()
d.bark()

print("\nMultilevel Inheritance:")
p = Puppy()
p.speak()
p.bark()
p.weep()

print("\nHierarchical Inheritance:")
c = Cat()
c.speak()
c.meow()

print("\nMultiple Inheritance:")
h = HybridAnimal()
h.speak()
h.bark()
h.friendly()
h.show()
```

```python
'''  MRO defines the order in which Python looks for a method or attribute when it is called on an object, especially in multiple / hybrid inheritance.

Python uses the C3 Linearization algorithm to compute MRO.

✔ In simple words:

When you call obj.method(), Python checks classes from left to right, following the MRO list, and uses the first match it finds.

What is C3 Linearization (C3 Serialization)?

C3 linearization is the algorithm Python uses to compute MRO (Method Resolution Order) in multiple and hybrid inheritance.



C3 linearization produces a single, consistent order of classes that:

Preserves local precedence order

Maintains parent-before-child rule

Avoids ambiguity (Diamond problem)

🔑 Why C3 is Needed

In multiple inheritance, a class can inherit the same method from multiple paths.

❌ Without C3 → ambiguous method calls
✅ With C3 → predictable and conflict-free order

🔹 C3 Rules (Very Important)

C3 linearization must satisfy three properties:

1️⃣ Local Precedence Order

Order of parents in class definition is respected.

class D(B, C):   # B is checked before C

2️⃣ Monotonicity

A subclass cannot change the order of its parents’ MRO.

👉 If B comes before C in parent MRO, it must remain so.

3️⃣ No Repetition

Each class appears only once in the MRO.


  '''

class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.mro())
```

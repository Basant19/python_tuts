```python
''' Polymorphism is an OOP concept where the same function or method name performs different actions based on the object or data type.
'''
```

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Polymorphism
animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

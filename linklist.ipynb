```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


def traverse(head):
    currentnode = head
    while currentnode is not None:   # can also use: while currentnode:
        print(currentnode.data, end=" -> ")
        currentnode = currentnode.next
    print("NULL")


# ---------- Taking user input ----------
n = int(input("Enter number of nodes: "))

head = None
current = None

for i in range(n):
    data = int(input(f"Enter value for node {i+1}: "))
    new_node = Node(data)

    if head is None:          # first node
        head = new_node
        current = new_node
    else:                     # remaining nodes
        current.next = new_node
        current = new_node # we can also use current = current.next


# ---------- Traverse Linked List ----------
traverse(head)
```

```python
#insert at beginning 

def insert_at_beginning(head, data):
    new_node = Node(data)
    new_node.next = head   # link new node to old head
    head = new_node        # update head
    return head


# ---------- Already Existing Linked List ----------
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

head = node1

print("Before insertion:")
traverse(head)

# ---------- Insert at Beginning ----------
head = insert_at_beginning(head, 5)

print("After insertion at beginning:")
traverse(head)
```

```python
#insert at nth position
def insert_at_nth(head, data, n):
    # Edge case 1: Invalid position
    if n <= 0:
        print("Invalid position")
        return head

    # Edge case 2: Insert at beginning (n = 1)
    if n == 1:
        new_node = Node(data)
        new_node.next = head
        return new_node

    # Traverse to (n-1)th node
    current = head
    count = 1

    while current is not None and count < n - 1:
        current = current.next
        count += 1

    # Edge case 3: Position out of range
    if current is None:
        print("Position out of range")
        return head

    # Normal case: Insert in between / at end
    new_node = Node(data)
    new_node.next = current.next
    current.next = new_node

    return head


# ---------- Already Existing Linked List ----------
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1

print("Before insertion:")
traverse(head)

# ---------- Insert at Nth Position ----------
head = insert_at_nth(head, 25, 3)   # insert 25 at position 3

print("After insertion at position 3:")
traverse(head)
```

```python
#insert in the middle of linked list
def insert_middle_optimized(head, data):
    # Edge case: empty list
    if head is None:
        return Node(data)

    # Edge case: single node
    if head.next is None:
        new_node = Node(data)
        head.next = new_node
        return head

    slow = head
    fast = head

    while fast.next and fast.next.next:
        #Insertion → stop before middle → use fast.next and fast.next.next
        slow = slow.next
        fast = fast.next.next

    new_node = Node(data)
    new_node.next = slow.next
    slow.next = new_node

    return head


print("Before insertion:")
traverse(head)

head = insert_middle_optimized(head, 27)

print("After optimized middle insertion:")
traverse(head)
```

```python
def delete_at_nth(head, n):
    # Edge case 1: empty list
    if head is None:
        print("List is empty. Nothing to delete.")
        return head

    # Edge case 2: invalid position
    if n <= 0:
        print("Invalid position.")
        return head

    # Edge case 3: delete first node (n = 1)
    if n == 1:
        return head.next

    current = head
    count = 1

    # Traverse to (n-1)th node
    while current.next and count < n - 1:
        current = current.next
        count += 1

    # Edge case 4: position out of range
    if current.next is None:
        print("Position out of range.")
        return head

    # Delete nth node
    current.next = current.next.next

    return head

head = node1

print("Before deletion:")
traverse(head)

# ---------- Delete at Nth Position ----------
head = delete_at_nth(head, 3)   # delete node at position 3

print("After deleting position 3:")
traverse(head)
```

```python
def delete_middle(head):
    # Edge case 1: empty list
    if head is None:
        print("List is empty. Nothing to delete.")
        return head

    # Edge case 2: only one node
    if head.next is None:
        return None

    slow = head
    fast = head
    prev = None   # to keep track of node before slow

    while fast and fast.next:
        #Deletion → reach exact  middle → use fast and fast.next
        prev = slow
        slow = slow.next
        fast = fast.next.next

    # slow is at middle, prev is before middle
    prev.next = slow.next

    return head


head = node1

print("Before deleting middle:")
traverse(head)

# ---------- Delete Middle ----------
head = delete_middle(head)

print("After deleting middle:")
traverse(head)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


def traverse_circular(head):
    # Edge case: empty list
    if head is None:
        print("List is empty")
        return

    current = head
    while True:
        print(current.data, end=" -> ")
        current = current.next

        if current == head:
            break

    print("(back to head)")


# ---------- Create Circular Linked List ----------
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3
node3.next = node1   # circular link

head = node1

# ---------- Traverse ----------
traverse_circular(head)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


def traverse_forward(head):
    current = head
    while current:
        print(current.data, end=" <-> ")
        current = current.next
    print("NULL")


def traverse_backward(tail):
    current = tail
    while current:
        print(current.data, end=" <-> ")
        current = current.prev
    print("NULL")


# ---------- Create Doubly Linked List ----------
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1

node2.next = node3
node3.prev = node2

head = node1
tail = node3

# ---------- Traversals ----------
print("Forward traversal:")
traverse_forward(head)

print("Backward traversal:")
traverse_backward(tail)
```

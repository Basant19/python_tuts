```python
''' 
WITHOUT __init__

(Manual initialization – user must remember to call a method)
 '''
class Student:
    def set_data(self, name, roll):
        self.name = name
        self.roll = roll

    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)

# object creation
s1 = Student()

# manual initialization (IMPORTANT)
s1.set_data("Basant", 101)

# using data
s1.display()
```

```python
s2 = Student()
s2.display()   # ❌ ERROR: name and roll not set
```

```python
''' __init__ is a constructor which allows automatic initialization 
(at object creation – user does not need to remember to call a method)
__init__ allow to initialize the object at the time of its creation.
  '''


class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll

    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)

# object creation + initialization together
s1 = Student("Basant", 101)

# using data
s1.display()
```

```python
#without __str__ method
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)
print(p1)
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)
```

```python
#2.matrix representation
def create_adjacency_matrix(vertices, edges):
    
    # Step 1: Create empty matrix
    matrix = []
    
    # Step 2: Create rows one by one
    for i in range(vertices):
        
        row = []   # Create empty row
        
        # Step 3: Fill row with 0s
        for j in range(vertices):
            row.append(0)#can use -1 also 
        
        # Step 4: Add row to matrix
        matrix.append(row)
    
    # Step 5: Fill edges
    for u, v in edges:
        matrix[u][v] = 1
        matrix[v][u] = 1   # Because undirected graph
    
    return matrix
```

```python
#driver code adjacency matrix 

adj_list = create_adjacency_matrix(vertices, edges)
print("\nAdjacency List:")
for index, neighbors in enumerate(adj_list):
    print(index, "->", neighbors)
```

# Traversing
1.BFS iterative and recursive
2.DFS iterative and recursive  

```python
#BFS – Iterative (Adjacency List directly  input given )

from collections import deque

def bfs_iterative(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    result = []

    while queue:
        node = queue.popleft()
        result.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return result

graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0],
    3: [1],
    4: [1]
}

print ('graph present \n',graph)
# Driver Code
print("BFS Iterative (when Adj List input ):", bfs_iterative(graph, 0))
```

```python
#BFS – Recursive (Adjacency List input directly given)

from collections import deque

def bfs_recursive_helper(graph, queue, visited, result):
    if not queue:
        return

    node = queue.popleft()
    result.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            visited.add(neighbor)
            queue.append(neighbor)

    bfs_recursive_helper(graph, queue, visited, result)


def bfs_recursive(graph, start):
    visited = set([start])
    queue = deque([start])
    result = []

    bfs_recursive_helper(graph, queue, visited, result)
    return result
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0],
    3: [1],
    4: [1]
}

print ('graph present \n',graph)

# Driver Code
print("BFS Recursive (Adj List directly given ):", bfs_recursive(graph, 0))
```

```python
#DFS – Iterative (Adjacency List directly given)

def dfs_iterative(graph, start):
    visited = set()
    stack = [start]
    result = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            result.append(node)

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return result

graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0],
    3: [1],
    4: [1]
}

print ('graph present \n',graph)

# Driver Code
print("DFS Iterative (Adj List directly given ):", dfs_iterative(graph, 0))
```

```python
# DFS – Recursive (Adjacency List directly given)

def dfs_recursive_helper(graph, node, visited, result):
    visited.add(node)
    result.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs_recursive_helper(graph, neighbor, visited, result)


def dfs_recursive(graph, start):
    visited = set()
    result = []

    dfs_recursive_helper(graph, start, visited, result)
    return result

graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0],
    3: [1],
    4: [1]
}

print ('graph present \n',graph)

# Driver Code
print("DFS Recursive (Adj List):", dfs_recursive(graph, 0))

```

```python
#When Input is EDGE LIST

#Convert Edge List → Adjacency List

def build_graph(edges, n):
    graph = {i: [] for i in range(n)}

    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)  # remove this line if directed

    return graph

edges = [(0,1), (0,2), (1,3), (1,4)]
n = 5
print ('given edges :',edges)
print ('given vertices:',n)

graph_from_edges = build_graph(edges, n)

print ('given graph after converting edge list to adjacency list ',graph_from_edges)
print("BFS Iterative (when graph after converting edge list to adjacency list ):", bfs_iterative(graph_from_edges, 0))
print("BFS Recursive (when graph after converting edge list to adjacency list):", bfs_recursive(graph_from_edges, 0))
print("DFS Iterative (when graph after converting edge list to adjacency list):", dfs_iterative(graph_from_edges, 0))
print("DFS Recursive (when graph after converting edge list to adjacency list):", dfs_recursive(graph_from_edges, 0))
```

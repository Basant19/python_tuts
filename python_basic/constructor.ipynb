```python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll

    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)

s1 = Student("Basant", 101)
s1.display()
```

```python
#super () function is used to give access to the methods and properties of a parent or sibling class.
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

e = Employee("Basant", 50000)
print(e.name, e.salary)
```
